```python
#SHOPPING BILL CALCULATOR
#by sonam shrivastav


items={
    "mango": {"price":40,"qty":3},
    "apple": {"price":30,"qty":4},
    "berry": {"price":50,"qty":5},
}
total_bill=0

for name,details in items.items():
    subtotal=details["price"]*details["qty"]
    total_bill +=subtotal
    print(f"{name.capitalize():10} rs{subtotal:.2f}")
discount=total_bill*0.10
final_bill=total_bill-discount
print(f"{'':10} -----------")
print(f"{'subtotal':10} rs{total_bill:.2f}")
print(f"{'discount':10} rs{discount:.2f}")
print(f"{'Total':10} rs{final_bill:.2f}")
```
