```python
#stored account details
account_holder = "Sonam jain"
account_number = "SB001234"
pin            = "5678"
balance        = 25000
account_type   = "Savings"
print("====Welcome to Python Bank ====")

# verigy login
user_acc = input("Enter account number : ")
user_pin = input("Enter PIN            : ")
print("===============================")
if user_acc == account_number and user_pin == pin:
    print(f"Welcome {account_holder}!")
    print(f"Account Type : {account_type}")
    print("===============================")
# show menu
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Account Details")
    print("5. Exit")
    print("==================================")
    choice = input("Enter your choice : ")
    if choice == "1":
        print(f"Balance : {balance}")
    elif choice == "2":
        deposit = int(input("Enter your money to be deposit : "))
        if deposit > 100:
            print("Deposit successful")
            print(f"New Balance : {balance+deposit}")
        else:
            print("minimum deposit amount RS100")

    elif choice == "3":
        withdraw = int(input("Enter withdrawal amount : "))
        print("====================================")
        if withdraw < 100:
            print("minimum withdrawal amount is RS100")
        elif withdraw > 10000:
            print("maximum withdrawal per transaction is Rs10000")
        elif withdraw > balance:
            print("Cannot exceed available balance")
        else:
            print(f"Remaining balance : {balance-withdraw}")
            print("================================")

    elif choice == "4":
        print(f"Account_holder           : {account_holder}")
        print(f"Account_number           : {account_number}")
        print(f"Available Balance        : {balance}")
        print(f"Account type             : {account_type}")

    elif choice == "5":
        print("Thank you for banking with us!")

    else:
        print("Invalid choice! Enter 1 to 5 only")

elif user_acc != account_number:
    print("Account not found!")
else:
    print("Wrong PIN!")
    print("===============================")


```
