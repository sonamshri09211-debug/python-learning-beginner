This project analyses student marks,calculates averages,assigns grades and summarizes class performance.

```python
# STUDENT GRADE ANALYZER

Students={
    "sonam":[80,70,60,50],
    "priya":[40,45,78,37],
    "amit": [56,67,48,77],
    "vinay":[45,56,79,36]
}

#subjects names
subjects=["Maths","Python","AI","ML"]
```

```python


print("Student data loaded successfully")
print(f"Total Students: {len(Students)}")
print(f"subjects: {','.join(subjects)}")
```

```python
#FUNCTION TO CALCULATE GRADE

def calculate_grade(average):
  if average >= 90:
    return "A+"
  elif average >= 80:
    return "A"
  elif average >= 70:
    return "B"
  elif average >= 60:
    return "C"
  elif average >= 50:
    return "D"
  else:
    return "F"
def calculate_average(marks):
  return sum(marks)/len(marks)

print("Functions created successfully")


```

```python
#ANALYSE EACH STUDENT

print("=" *55)
print("STUDENT GRADE REPORT CARD")
print("=" *55)

all_averages=[]

for name,marks in Students.items():
  average=calculate_average(marks)
  grade=calculate_grade(average)
  all_averages.append(average)

  print(f"\n student: {name}")
  print(f"Marks: {marks}")
  print(f"Average: {average:.1f}/100")
  print(f"Grade: {grade}")
  print("-" *40)
```

```python
#CLASS SUMMARY

class_average=calculate_average(all_averages)
higest=max(all_averages)
lowest=min(all_averages)

#Find topper and lowest scorer
names=list(Students.keys())
topper=names[all_averages.index(higest)]
needs_help=names[all_averages.index(lowest)]

print("\n "+"="*55)
print("class performance summary")
print("="*55)
print(f"Total Students:{len(Students)}")
print(f"Class Average:{class_average:.1f}/100")
print(f"Higest score:{higest:.1f}({topper})")
print(f"Lowest score:{lowest:.1f}({needs_help})")
print(f"Class Topper:{topper}")
print("="*55)
print("\n Analysis Complete")
```
